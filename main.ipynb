```python
import pandas as pd
import pyinputplus as pyip
print("ok")
```

```python
try:
  df = pd.read_csv('budget.csv')
  print("old file loaded")
except FileNotFoundError:
  df = pd.DataFrame(columns=['date','description','type','amount'])
  print("New empty table created")
```

```python
def calculate_balance(data):
  income = data[data['type']=='income']['amount'].sum()
  expense = data[data['type']=='expence']['amount'].sum()
  return income-expense
```

```python
test_df = pd.DataFrame([
    {'date': '2026-01-01', 'description': 'Salary', 'type': 'income', 'amount': 1000},
    {'date': '2026-01-02', 'description': 'Food', 'type': 'expense', 'amount': 150}
])
assert calculate_balance(test_df) == 850
print("Test 1 Passed - Balance is 850")

empty_df = pd.DataFrame(columns=['date', 'description', 'type', 'amount'])
assert calculate_balance(empty_df) == 0
print("Test 2 Passed - Empty balance is 0")
```

```python
def add_transaction(data):
    date = pyip.inputDate(prompt="Enter date (YYYY-MM-DD): ")
    description = pyip.inputStr(prompt="Enter description: ")
    type_val = pyip.inputChoice(['income', 'expense'], prompt="Enter type (income/expense): ")
    amount = pyip.inputFloat(prompt="Enter amount: ", min=0.01)

    new_row = pd.DataFrame([{'date': date, 'description': description, 'type': type_val, 'amount': amount}])
    new_data = pd.concat([data, new_row], ignore_index=True)
    print("Transaction Added!")
    return new_data
```

```python
def view_transactions(data):
    if data.empty:
        print("No transactions yet.")
    else:
        print(data)

def show_menu():
    print("\n--- Budget Manager ---")
    print("1. Add Transaction")
    print("2. View All Transactions")
    print("3. View Current Balance")
    print("4. Save & Exit")
```

```python
while True:
    show_menu()
    choice = pyip.inputChoice(['1', '2', '3', '4'], prompt="Choose option (1-4): ")

    if choice == '1':
        df = add_transaction(df)
    elif choice == '2':
        view_transactions(df)
    elif choice == '3':
        balance = calculate_balance(df)
        print(f"Current Balance: ${balance:.2f}")
    elif choice == '4':
        df.to_csv('budget.csv', index=False)
        print("Goodbye!")
        break
```
